# Searching for dark star candidates in JWST MIRI
What I've done so far and the difficulties I've encountered :]

## Data acquisition
- NIRSpec targets with z>9 and redshift grade=3 (downloaded from DJA)
- Sky location matches between DJA sources and MIRI entries (queried from MAST)
    - 24 MIRI spectrometry entries
    - 42 MIRI photometry entries
- x1d and i2d `.fit` files for all the matches


## SMDS / SMDS+galaxy spectroscopic fit
Data tend to be very noisy, especially at long wavelengths (the fit is inverse-error-weighted). Many spectra are noise-dominated throughout, causing the fitted SMDS mass and/or galaxy mass to sit at prior boundaries, like this one:

![noisy_spec_example](production/miri_joint_fit/plots/abell2744-castellano2-v4_17724_joint_fit.png)

All the spectra that don't look like pure noise seem to have galaxy emission features. This is one of the healthiest looking fits, with $\log(M_{\mathrm{SMDS}})=6.1$, $\log(M_{\mathrm{galaxy}})=7.6$:

![spec_fit_example](production/miri_joint_fit/plots/jades-gdn09-v4_55757_joint_fit.png)

Some spectra just look like this - a small stretch of noise at long wavelengths. I wonder where they come from.

![weird_spec_example](production/miri_joint_fit/plots/goodsn-wide0-v4_1268_joint_fit.png)
